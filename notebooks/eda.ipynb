# Exploratory Data Analysis — Tweet Sentiment

Quick look at the NLTK `twitter_samples` dataset used for training:
class balance, tweet-length distribution, and most frequent words per class.

```python
import nltk
nltk.download("twitter_samples", quiet=True)
from nltk.corpus import twitter_samples
import matplotlib.pyplot as plt
from collections import Counter
import sys
sys.path.insert(0, "..")
from src.preprocess import clean_text
```

```python
pos = [clean_text(t) for t in twitter_samples.strings("positive_tweets.json")]
neg = [clean_text(t) for t in twitter_samples.strings("negative_tweets.json")]
pos = [t for t in pos if t]
neg = [t for t in neg if t]
print(f"positive: {len(pos)}, negative: {len(neg)}")

plt.bar(["positive", "negative"], [len(pos), len(neg)])
plt.title("Class balance")
plt.ylabel("tweets")
plt.show()
```

```python
pos_len = [len(t.split()) for t in pos]
neg_len = [len(t.split()) for t in neg]
plt.hist([pos_len, neg_len], bins=20, label=["positive", "negative"], alpha=0.7)
plt.title("Tweet length distribution (words)")
plt.xlabel("words")
plt.legend()
plt.show()
```

```python
def top_words(tweets, n=15):
    return Counter(w for t in tweets for w in t.split()).most_common(n)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, (title, tweets) in zip(axes, [("positive", pos), ("negative", neg)]):
    words, counts = zip(*top_words(tweets))
    ax.barh(words[::-1], counts[::-1])
    ax.set_title(f"Top words — {title}")
plt.tight_layout()
plt.show()
```
